```python
import numpy as np
import plotly.graph_objects as go
import arbol_kd as akd
```

```python
def show_search_result_2d(points, query, result):
    fig = go.Figure()
    fig.add_trace(
        go.Scatter(
            x=points[:, 0],
            y=points[:, 1],
            mode="markers",
            name="Points",
            marker=dict(size=5, opacity=0.5),
        )
    )
    fig.add_trace(
        go.Scatter(
            x=result[:, 0],
            y=result[:, 1],
            mode="markers",
            name="Search Result",
            marker=dict(size=5, opacity=0.5),
        )
    )
    fig.add_trace(
        go.Scatter(
            x=[query[0]],
            y=[query[1]],
            mode="markers",
            name="Query Point",
            marker=dict(symbol="x-thin", size=20, line_width=4, line_color="orange"),
        )
    )
    fig.update_layout(
        title="KD-Tree Search Result",
        xaxis_title="X-axis",
        yaxis_title="Y-axis",
        showlegend=True,
        width=600,
        height=600,
    )
    fig.show()


points_2d = np.random.rand(10000, 2)  # generate points in range [0, 1]
tree_2d = akd.build_tree_2d(points_2d)  # build a KD-tree
box_2d = np.array([[0, 1], [0, 1]])  # define a box in 2D space size [0, 1] x [0, 1]
```

```python
query = np.array([0.5, 0.5])  # define a query point
result = akd.k_nearest_periodic(tree_2d, query, box_2d, k=1000)

show_search_result_2d(points_2d, query, result)

query = np.array([0.95, 0.95])
result = akd.k_nearest_periodic(tree_2d, query, box_2d, k=1000)
show_search_result_2d(points_2d, query, result)
```

```python
def show_search_result_3d(points, query, result):
    fig = go.Figure(
        data=[
            go.Scatter3d(
                x=points[:, 0],
                y=points[:, 1],
                z=points[:, 2],
                mode="markers",
                marker=dict(size=5, opacity=0.2),
                name="Points",
            )
        ]
    )

    fig.add_trace(
        go.Scatter3d(
            x=result[:, 0],
            y=result[:, 1],
            z=result[:, 2],
            mode="markers",
            marker=dict(size=5, opacity=0.2),
            name="Search Result",
        )
    )

    fig.add_trace(
        go.Scatter3d(
            x=[query[0]],
            y=[query[1]],
            z=[query[2]],
            mode="markers",
            marker=dict(symbol="x", size=10, color="orange"),
            name="Query Point",
        )
    )
    fig.update_layout(
        title="KD-Tree Search Result in 3D",
        scene=dict(xaxis_title="X-axis", yaxis_title="Y-axis", zaxis_title="Z-axis"),
        width=600,
        height=600,
    )
    fig.show()


points_3d = np.random.rand(10000, 3)  # generate points in range [0, 1]
tree_3d = akd.build_tree_3d(points_3d)  # build a KD-tree
box_3d = np.array(
    [[0, 1], [0, 1], [0, 1]]
)  # define a box in 3D space size [0, 1] x [0, 1] x [0, 1]
```

```python
query = np.array([0.5, 0.5, 0.5])  # define a query point
result = akd.k_nearest_periodic(tree_3d, query, box_3d, k=1000)

show_search_result_3d(points_3d, query, result)

query = np.array([0.95, 0.95, 0.95])
result = akd.k_nearest_periodic(tree_3d, query, box_3d, k=1000)
show_search_result_3d(points_3d, query, result)
```

```python
def show_nearest_3d(result, query):
    fig = go.Figure()
    fig.add_trace(
        go.Scatter3d(
            x=result[:, 0],
            y=result[:, 1],
            z=result[:, 2],
            mode="markers",
            marker=dict(size=5, opacity=0.2),
            name="Image Points",
        )
    )
    fig.add_trace(
        go.Scatter3d(
            x=[query[0]],
            y=[query[1]],
            z=[query[2]],
            mode="markers",
            name="Query Point",
            marker=dict(symbol="x", size=10, color="orange"),
        )
    )
    fig.update_layout(
        title="Nearest Points Search Result",
        scene=dict(xaxis_title="X-axis", yaxis_title="Y-axis", zaxis_title="Z-axis"),
        showlegend=True,
        width=600,
        height=600,
    )
    fig.show()
```

```python
query = np.array([0.95, 0.95, 0.95])
result = akd.k_nearest_periodic(tree_3d, query, box_3d, k=1000)
result_image = akd.clamp2nearest(result, query, box_3d)
show_nearest_3d(result_image, query)
```
